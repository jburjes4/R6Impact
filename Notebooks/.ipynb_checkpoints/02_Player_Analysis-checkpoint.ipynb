```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine, text
import matplotlib.pyplot as plt
```

```python
engine = create_engine(
    "mysql+pymysql://root:JapanTokyoSpain2025!@localhost/r6impact"
)
```

```python
query = """
SELECT *
FROM kill_events
WHERE NOT (killer = victim);
"""

df = pd.read_sql(query, engine)
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
player_kills = (
    df.groupby("killer")
    .size()
    .reset_index(name="kills")
    .sort_values("kills", ascending=False)
)
```

```python
player_kills
```

```python
player_deaths = (
    df.groupby("victim")
      .size()
      .reset_index(name="deaths")
      .sort_values("deaths", ascending=False)
)
```

```python
player_kills = player_kills.rename(columns={"killer": "player"})
player_deaths = player_deaths.rename(columns={"victim": "player"})


player_kd = player_kills.merge(
    player_deaths,
    on="player",
    how="outer"
)


# Replace missing values with 0
player_kd[["kills", "deaths"]] = (
    player_kd[["kills", "deaths"]].fillna(0)
)


player_kd["K/D"] = (player_kd["kills"] - player_kd["deaths"]
)

player_kd = player_kd.sort_values(
    "K/D",
    ascending=False
)

player_kd
```

```python
player_kd.to_csv(
    "../Exports/player_kd.csv",
    index=False
)
```

```python
player_entry_kills = (
    df[df["is_first_kill"] == 1]
    .groupby("killer")
    .size()
    .reset_index(name="entry_kills")
    .sort_values("entry_kills", ascending=False)
)

player_entry_kills.head(20)
```

```python
player_entry_deaths = (df[df["is_first_kill"] == 1]
    .groupby("victim")
    .size()
    .reset_index(name="entry_deaths")
    .sort_values("entry_deaths", ascending=False)
)

player_entry_deaths.head(20)
```

```python
player_entry_kills = player_entry_kills.rename(columns={"killer": "player"})
player_entry_deaths = player_entry_deaths.rename(columns={"victim": "player"})


player_entry_net = player_entry_kills.merge(
    player_entry_deaths,
    on="player",
    how="outer"
)

player_entry_net[["entry_kills", "entry_deaths"]] = (
    player_entry_net[["kills", "deaths"]].fillna(0)
)

player_entry_net["net"] = (player_entry_kills["entry_kills"] - player_entry_deaths["entry_deaths"]
)

player_entry_net = player_entry_net.sort_values(
    "net",
    ascending=False
)

player_entry_net
```
